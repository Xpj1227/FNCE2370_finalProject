```python
import numpy as np


def portfolio_return(weights, asset_returns):
    """
    Compute one-period portfolio return.

    Parameters
    ----------
    weights : array-like of shape (n_assets,)
        Portfolio weights.
    asset_returns : array-like of shape (n_assets,)
        Realized asset returns for the same period.

    Returns
    -------
    float
        Realized portfolio return w'r.
    """
    w = np.asarray(weights, dtype=float)
    r = np.asarray(asset_returns, dtype=float)

    if w.ndim != 1 or r.ndim != 1:
        raise ValueError("weights and asset_returns must both be 1D arrays")
    if w.shape[0] != r.shape[0]:
        raise ValueError("weights and asset_returns must have the same length")

    return float(np.dot(w, r))



def mean_variance_reward(weights, asset_returns, gamma):
    """
    Mean-variance style one-period reward used by the Q-learning agent:

        R_{t+1} = r_{p,t+1} - (gamma / 2) * r_{p,t+1}^2

    Parameters
    ----------
    weights : array-like of shape (n_assets,)
        Portfolio weights chosen at time t.
    asset_returns : array-like of shape (n_assets,)
        Realized asset returns at time t+1.
    gamma : float
        Risk-aversion parameter.

    Returns
    -------
    float
        One-period reward.
    """
    if gamma < 0:
        raise ValueError("gamma must be non-negative")

    rp = portfolio_return(weights, asset_returns)
    reward = rp - 0.5 * gamma * (rp ** 2)
    return float(reward)



def reward_from_portfolio_return(portfolio_ret, gamma):
    """
    Compute the reward directly from a realized portfolio return.
    Useful when portfolio return has already been computed elsewhere.
    """
    rp = float(portfolio_ret)
    if gamma < 0:
        raise ValueError("gamma must be non-negative")
    return float(rp - 0.5 * gamma * (rp ** 2))


# Example usage
if __name__ == "__main__":
    # Example: SPY, IEF, GLD
    w = np.array([0.5, 0.3, 0.2])
    r_next = np.array([0.01, -0.002, 0.004])
    gamma = 3.0

    rp = portfolio_return(w, r_next)
    reward = mean_variance_reward(w, r_next, gamma)

    print(f"Portfolio return: {rp:.6f}")
    print(f"Reward: {reward:.6f}")
```
